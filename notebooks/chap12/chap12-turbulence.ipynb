```python
omega = 1
t = np.arange(0, 2*np.pi, 0.1)

phi_values = [0, np.pi / 4, np.pi / 2]
titles = [r'$\phi = 0$', r'$\phi = \frac{\pi}{4}$', r'$\phi = \frac{\pi}{2}$']

# Create 3 subplots in a row
fig, axes = plt.subplots(1, 3, figsize=(15, 4), sharex=True, sharey=True)

for ax, phi, title in zip(axes, phi_values, titles):
    u = np.cos(omega * t)
    v = np.cos(omega * t + phi)
    
    ax.scatter(u, v, s=20, facecolors=FG, edgecolors=FG, zorder=2, clip_on=False)
    
    ax.set_xticks(np.arange(-1, 1.1, 1))
    ax.set_yticks(np.arange(-1, 1.1, 1))
    ax.tick_params(axis='both', labelsize=14)
    
    ax.set_xlabel(r'$u(t)$', fontsize=18)
    ax.set_title(title, fontsize=22, pad=15)
    ax.minorticks_off()

# Add a shared y-axis label
axes[0].set_ylabel(r'$v(t)$', fontsize=18)

plt.minorticks_off()
savefig(plt.gcf(), 'chap12ex12.4.eps', bbox_inches=None)
plt.show()
```
